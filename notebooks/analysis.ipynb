```python
import json

import numpy as np
import pandas as pd
from utils.data import Cst

with open("../trees.json", "r") as file:
    data = [Cst.from_dict(sample) for sample in json.load(file)]
```

# Number of Samples by Language

```python
df = pd.DataFrame([{"language": sample.language} for sample in data])

samples_per_language = []

for language in df["language"].unique():
    subset = df[df["language"] == language]
    samples_per_language.append(
        {"Language": language, "Number of Functions": len(subset)}
    )

samples_per_language.append({"Language": "Total", "Number of Functions": len(df)})
pd.DataFrame(samples_per_language)
```

# Leaf Count 

```python
df = pd.DataFrame(
    [{"language": sample.language, "num_leafs": sample.num_leaves} for sample in data]
)

leaf_count_by_language = []

for language in df["language"].unique():
    subset = df[df["language"] == language]
    leaf_counts = subset["num_leafs"]
    mean = leaf_counts.mean()
    stdev = leaf_counts.std()
    min = leaf_counts.min()
    max = leaf_counts.max()
    med = leaf_counts.median()
    mode = leaf_counts.mode().item()

    leaf_count_by_language.append(
        {
            "Language": language,
            "Mean": mean,
            "Standard Deviation": stdev,
            "Min": min,
            "Max": max,
            "Median": med,
            "Mode": mode,
            "95% Percentile": np.percentile(leaf_counts, 95),
        }
    )

leaf_count_by_language.append(
    {
        "Language": "Total",
        "Mean": df["num_leafs"].mean(),
        "Standard Deviation": df["num_leafs"].std(),
        "Min": df["num_leafs"].min(),
        "Max": df["num_leafs"].max(),
        "Median": df["num_leafs"].median(),
        "Mode": df["num_leafs"].mode().item(),
        "95% Percentile": np.percentile(df["num_leafs"], 95),
    }
)

pd.DataFrame(leaf_count_by_language)
```

# Leaf Node Depth

```python
tmp = []

for sample in data:
    v, c = np.unique(sample.depth[sample.leaf_indicies], return_counts=True)
    mode = v[np.argmax(c)]
    tmp.append(
        {
            "language": sample.language,
            "max": sample.depth[sample.leaf_indicies].max(),
            "mean": sample.depth[sample.leaf_indicies].mean(),
            "std": sample.depth[sample.leaf_indicies].std(),
            "median": np.median(sample.depth[sample.leaf_indicies]),
            "mode": mode,
        }
    )

tmp = pd.DataFrame(tmp)
tmp
```
